```python
import numpy as np
```

```python
products = ["A", "B", "C", "D", "E"]
sales_2023 = [10000, 15000, 12000, 18000, 20000]
sales_2024 = [12000, 17000, 14000, 20000, 22000]
```

```python
x=np.arange(len(products))
width=0.35
```

```python
x
```

```python
plt.bar(x-width/2,sales_2023,width,label="2023")
plt.bar(x+width/2,sales_2024,width,label="2024")
plt.title("Products vs Sales")
plt.xlabel("Products")
plt.ylabel("Sales")
plt.legend()
plt.show()
```

#Stacked bar chart

```python
categories = ["A", "B", "C", "D", "E"]
online_sales = [10000, 15000, 12000, 18000, 20000]
offline_sales = [8000, 12000, 10000, 15000, 18000]
```

```python
plt.bar(categories,online_sales,label="Online Sales")
plt.bar(categories,offline_sales,bottom=online_sales,label="Offline Sales")

plt.title("Products vs Sales")
plt.legend()
plt.show()
```

```python
plt.barh(x-width/2,sales_2023,width,label="2023")
plt.barh(x+width/2,sales_2024,width,label="2024")
plt.title("Products vs Sales")
plt.xlabel("Sales")
plt.ylabel("Products")
plt.legend()
plt.show()
```

#Radial chart

```python
df1 = pd.DataFrame({
    "Skill": ["Python", "SQL", "Excel", "Statistics", "Machine Learning"],
    "Score": [90, 85, 80, 75, 70]
})
```

```python
labels = df1["Skill"].tolist()
values = df1["Score"].tolist()

angles = np.linspace(0,2*np.pi,len(labels),endpoint = False)

values = np.concatenate((values,[values[0]]))
angles = np.concatenate((angles,[angles[0]]))

plt.polar(angles,values)
plt.title("Radial plot")
plt.show()
```

#Radar Chart

```python
labels = df1["Skill"].tolist()
values = df1["Score"].tolist()
angles = np.linspace(0,2*np.pi,len(labels),endpoint = False)

values = np.concatenate((values,[values[0]]))
angles = np.concatenate((angles,[angles[0]]))

plt.polar(angles,values)
plt.fill(angles,values,alpha = 0.2)
```
